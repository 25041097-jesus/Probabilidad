```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Carga del dataset y limpieza inicial
df = pd.read_csv('autos2.csv')
precios = df['price'].dropna()
sns.set_theme(style="whitegrid")
```

```python
minimo = precios.min()
print("Valor mínimo:", minimo)
```

```python
maximo = precios.max()
print("Valor máximo:", maximo)
```

```python
moda = precios.mode().values
print("Moda(s):", moda)
```

```python
media = precios.mean()
print("Media:", round(media, 2))
```

```python
mediana = precios.median()
print("Mediana:", mediana)
```

```python
desviacion = precios.std()
print("Desviación estándar:", round(desviacion, 2))
```

```python
print(precios.describe())
```

```python
n = len(precios)
k_sturges = int(np.ceil(1 + 3.322 * np.log10(n)))
print("Bins según Sturges:", k_sturges)
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(precios, bins=k_sturges, kde=True, color='skyblue')
plt.title('Histograma de Precios de Automóviles (Regla de Sturges)')
plt.xlabel('Precio (USD)')
plt.ylabel('Frecuencia (Número de Autos)')
plt.tight_layout()
plt.show()
```

**Aplicando la fórmula $K = 1 + 3.322 \cdot \log_{10}(201) = 8.64$, redondeamos al entero superior para obtener $9$ contenedores óptimos que representen la distribución sin perder detalle.**

```python
plt.figure(figsize=(8, 5))
sns.boxplot(x='body-style', y='price', data=df)
plt.title('Relación entre el Precio del Automóvil y el Tipo de Vehículo')
plt.xlabel('Tipo de Vehículo (Body Style)')
plt.ylabel('Precio (USD)')
plt.tight_layout()
plt.show()
```

**Asimetría positiva: La mayor concentración de datos está sesgada hacia la izquierda, ubicándose entre los  $5,000  y  $12,000 .Caída progresiva: Conforme el precio aumenta hacia los  $20,000  y  $45,000 , la cantidad de vehículos disminuye drásticamente. Esto confirma que el mercado consta principalmente de autos de volumen/económicos y una minoría de gama alta.**

```python
plt.figure(figsize=(8, 5))
sns.boxplot(x='drive-wheels', y='price', data=df)
plt.title('Relación entre el Precio del Automóvil y el Tipo de Tracción')
plt.xlabel('Tipo de Tracción (Drive Wheels)')
plt.ylabel('Precio (USD)')
plt.tight_layout()
plt.show()
```

T**racción Delantera (fwd) y 4x4 (4wd): Mantiene precios reducidos y estables (medianas de $\$8,192.00$ y $\$9,005.50$).Tracción Trasera (rwd): Presenta una mediana considerablemente más alta ($\$16,900.00$) y abarca los precios más elevados del dataset, asociándose directamente a vehículos deportivos y ejecutivos.** **texto en negrita**

```python
plt.figure(figsize=(8, 5))
sns.boxplot(x='engine-location', y='price', data=df)
plt.title('Relación entre el Precio del Automóvil y la Ubicación del Motor')
plt.xlabel('Ubicación del Motor (Engine Location)')
plt.ylabel('Precio (USD)')
plt.tight_layout()
plt.show()
```

**Motor Frontal (front): Concentra a casi la totalidad de los autos ($198$ de $201$), abarcando un rango amplio que va desde el precio mínimo ($\$5,118.00$) hasta el máximo.Motor Trasero (rear): Representa una categoría reducida de solo $3$ autos, pero con un rango de costo exclusivo y elevado (entre $\$32,528.00$ y $\$37,028.00$).**

***conclusiones:***

**Comportamiento Asimétrico del Mercado: La variable precio presenta una asimetría positiva muy marcada. Mientras que la mediana se sitúa en $\$10,295.00$ y abarca el volumen masivo de vehículos utilitarios y de uso diario, el promedio se eleva hasta los $\$13,207.13$ debido a un grupo reducido pero significativo de modelos de gama alta que alcanzan hasta los $\$45,400.00$.Segmentación por Ingeniería: Las características mecánicas de los vehículos actúan como los principales clasificadores de valor:Tracción y Ubicación del Motor: La tracción delantera (fwd) y el motor frontal (front) dominan los segmentos de entrada y costo accesible. En cambio, la tracción trasera (rwd) y la ubicación del motor en la parte posterior (rear) se reservan casi exclusivamente para vehículos deportivos o de lujo.Estilo de Carrocería: Los modelos hatchback y sedan representan el mercado comercial de gran volumen, mientras que los diseños convertible y hardtop encabezan la gama de precios elevados.Valor del Análisis Exploratorio: Aplicar herramientas visuales como el histograma configurado bajo la regla de Sturges ($9$ intervalos) y los diagramas de caja (boxplots) permite identificar rápidamente la concentración de los datos, la presencia de valores atípicos (outliers) y las variables mecánicas que más impactan en la valoración económica de un auto.**
