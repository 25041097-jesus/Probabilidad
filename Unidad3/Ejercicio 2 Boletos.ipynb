```python
import numpy as np
import matplotlib.pyplot as plt

premios = ['A', 'B', 'C', 'D']
monto = np.array([0, 10, 20, 50])
p = np.array([0.50, 0.30, 0.15, 0.05])

suma_prob = np.sum(p)
print(f"Suma de probabilidades: {suma_prob:.2f}")

esperanza = np.sum(monto * p)
esperanza_x2 = np.sum((monto ** 2) * p)
varianza = esperanza_x2 - (esperanza ** 2)
desviacion = np.sqrt(varianza)

print(f"Ganancia Promedio Esperada E(X): ${esperanza:.2f} pesos")
print(f"Varianza (σ²): {varianza:.2f}")
print(f"Desviación Estándar (σ): ${desviacion:.2f} pesos")

acumulada = np.cumsum(p)
print("\nDistribución Acumulada F(x):")
for m, Fxi in zip(monto, acumulada):
    print(f"F({m}) = {Fxi:.2f}")

plt.figure(figsize=(8, 5))
plt.bar(monto, p, color='lightgreen', edgecolor='black', width=4, alpha=0.8, label='P(X)')
plt.grid(True, linestyle='--', alpha=0.6)

plt.axvline(x=esperanza, color='red', linestyle='--', label=f"Media E(X) = ${esperanza:.2f}")

plt.axvspan(max(0, esperanza - desviacion), esperanza + desviacion, color='red', alpha=0.15,
            label=f"±1 Desv. Estándar (σ = ${desviacion:.2f})")

plt.title("Distribución de Probabilidad - Premios de la Tómbola", fontsize=12, fontweight='bold')
plt.xlabel("Monto del premio en pesos (X)", fontsize=10)
plt.ylabel("Probabilidad P(X)", fontsize=10)

plt.legend()
plt.ylim(0, max(p) + 0.05)
plt.xticks(monto)
plt.show()
```

¿Qué significa el valor esperado en este contexto?
Significa que, si se juega muchas veces en la tómbola, la ganancia promedio por boleto a largo plazo será de $8.50 pesos.

Depende del precio del boleto:

 Si el boleto cuesta más de 8.50 pesos, el juego es desfavorable (la casa o la tómbola gana en promedio a largo plazo).
