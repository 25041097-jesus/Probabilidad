```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([0, 1, 2, 3, 4, 5])
p = np.array([0.10, 0.25, 0.30, 0.20, 0.10, 0.05])


suma_prob = np.sum(p)
print(f"Suma de probabilidades: {suma_prob:.2f}")

esperanza = np.sum(x * p)
esperanza_x2 = np.sum((x ** 2) * p)
varianza = esperanza_x2 - (esperanza ** 2)
desviacion = np.sqrt(varianza)

print(f"Valor Esperado E(X): {esperanza:.2f}")
print(f"Varianza (σ²): {varianza:.2f}")
print(f"Desviación Estándar (σ): {desviacion:.2f}")

acumulada = np.cumsum(p)
print("\nDistribución Acumulada F(x):")
for xi, Fxi in zip(x, acumulada):
    print(f"F({xi}) = {Fxi:.2f}")

plt.figure(figsize=(8, 5))
plt.bar(x, p, color='skyblue', edgecolor='black', alpha=0.7, label='P(X)')
plt.grid(True, linestyle='--', alpha=0.6)

plt.axvline(x=esperanza, color='red', linestyle='--', label=f"Media (E[X]) = {esperanza:.2f}")

plt.axvspan(esperanza - desviacion, esperanza + desviacion, color='red', alpha=0.2,
            label=f"±1 Desv. Estándar (σ = {desviacion:.2f})")
plt.title("Distribución de Probabilidad - Células Anormales", fontsize=12, fontweight='bold')
plt.xlabel("Número de células anormales (X)", fontsize=10)
plt.ylabel("Probabilidad P(X)", fontsize=10)
plt.legend()
plt.ylim(0, max(p) + 0.05)
plt.xticks(x)
plt.show()
```
