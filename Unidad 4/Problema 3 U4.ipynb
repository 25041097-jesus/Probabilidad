Tipo de solución: Es de tipo Poisson.  

¿Por qué?: El problema especifica que los eventos ocurren en un intervalo de tiempo continuo (un día) con una tasa promedio constante (lambda = 5) de forma independiente.


```python
from scipy.stats import poisson

# Parámetros de la distribución de Poisson
media = 5
X = 2  # Número de personas ingresadas

# Calcular la probabilidad de que el número de personas sea 2
probabilidad_dos_personas = poisson.pmf(X, media)

print("La probabilidad de que el número de personas ingresadas sea dos es:", probabilidad_dos_personas)


from scipy.stats import poisson

# Parámetros de la distribución de Poisson
media = 5
X = 10  # Límite de personas ingresadas

# Calcular la probabilidad de que X exceda de 10 -> P(X > 10) = 1 - P(X <= 10)
probabilidad_exceda_diez = 1 - poisson.cdf(X, media)

print("La probabilidad de que el número de personas exceda de 10 es:", probabilidad_exceda_diez)
```

Explicación: No es probable que x exceda de 10, ya que la probabilidad calculada es de aproximadamente 0.0137 (un 1.37%), lo cual representa un evento poco frecuente dado que el promedio de ingresos es de 5 personas por día
