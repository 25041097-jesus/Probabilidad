```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Problema 1: Estudiantes que pesan entre 68 y 72 kg

plt.figure(figsize=(6, 4))

# Parámetros de la distribución normal
media = 70            # Media
desv_std = 4          # Desviación estándar
total_estudiantes = 500

# Límites de peso
peso_min = 68
peso_max = 72

# Calcular las probabilidades acumuladas (CDF) para ambos límites
prob_hasta_72 = norm.cdf(peso_max, media, desv_std)
prob_hasta_68 = norm.cdf(peso_min, media, desv_std)

# La probabilidad entre ambos valores es la diferencia
prob_entre_68_72 = prob_hasta_72 - prob_hasta_68

# Cálculo del número de estudiantes en ese rango
estudiantes_entre_68_72 = total_estudiantes * prob_entre_68_72

print("Probabilidad de que un estudiante pese entre 68 y 72 kg:", prob_entre_68_72)
print("Número aproximado de estudiantes que pesan entre 68 y 72 kg:", estudiantes_entre_68_72)

# Graficar la distribución normal y sombrear el área entre 68 y 72 kg
x = np.linspace(media - 4 * desv_std, media + 4 * desv_std, 1000)
y = norm.pdf(x, media, desv_std)

plt.plot(x, y, 'b', linewidth=2)
plt.fill_between(x, y, where=((x >= peso_min) & (x <= peso_max)), color='skyblue', alpha=0.5)
plt.title('Probabilidad entre 68 kg y 72 kg. \n $\\mu=70$, $\\sigma=4$, X=500')
plt.xlabel('Peso (kg)')
plt.ylabel('Densidad de probabilidad')
plt.grid(True)
plt.show()
```
