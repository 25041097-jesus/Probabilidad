Tipo de solución: Es de tipo Binomial.

¿Por qué?: Porque se analiza un número fijo de ensayos independientes (n = 1000 motores), donde cada motor tiene una probabilidad constante de ser defectuoso (p = 0.001), y se busca el número de éxitos (motores defectuosos) en el total del lote. (Nota: Debido a que n es grande y p es pequeño, también se podría modelar mediante la aproximación de Poisson con lambda = np = 1, pero la naturaleza del problema parte de un proceso binomial).



```python
from scipy.stats import binom

# Parámetros de la distribución binomial
num_motores = 1000
probabilidad_defectuoso = 0.001
X = 0  # Ninguno defectuoso

# Calcular la probabilidad de que ningún motor sea defectuoso
probabilidad_ninguno = binom.pmf(X, num_motores, probabilidad_defectuoso)

print("La probabilidad de que ninguno sea defectuoso es:", probabilidad_ninguno)

from scipy.stats import binom

# Parámetros de la distribución binomial
num_motores = 1000
probabilidad_defectuoso = 0.001
X = 3  # Tres defectuosos

# Calcular la probabilidad de que tres motores sean defectuosos
probabilidad_tres = binom.pmf(X, num_motores, probabilidad_defectuoso)

print("La probabilidad de que tres sean defectuosos es:", probabilidad_tres)


from scipy.stats import binom

# Parámetros de la distribución binomial
num_motores = 1000
probabilidad_defectuoso = 0.001
X = 4  # Cuatro defectuosos

# Calcular la probabilidad de que cuatro motores sean defectuosos
probabilidad_cuatro = binom.pmf(X, num_motores, probabilidad_defectuoso)

print("La probabilidad de que cuatro sean defectuosos es:", probabilidad_cuatro)

```
